```python
def isMatch(s: str, p: str) -> bool:
    m, n = len(s), len(p)

    # dp[i][j] = does s[i:] match p[j:] ?
    dp = [[False] * (n + 1) for _ in range(m + 1)]

    # Base case:
    # Empty string matches empty pattern
    dp[m][n] = True

    # Fill table from the BACK (right-bottom) to the FRONT (top-left)
    for i in range(m, -1, -1):        # from m down to 0
        for j in range(n - 1, -1, -1):  # from n-1 down to 0

            # Check if current characters match (only if s still has characters)
            match = (i < m) and (p[j] == s[i] or p[j] == ".")

            # If the next pattern char exists AND is '*'
            if j + 1 < n and p[j + 1] == "*":

                # OPTION 1:
                # Skip the "X*" entirely → treat it as empty
                # So we jump over X*
                skip_star = dp[i][j + 2]

                # OPTION 2:
                # Use one occurrence of X* → only possible if characters match
                # If we use X*, we consume ONE char from s, but stay at same pattern position j
                use_star = match and dp[i + 1][j]

                dp[i][j] = skip_star or use_star

            else:
                # Normal matching rule:
                # Characters must match AND the rest must match
                dp[i][j] = match and dp[i + 1][j + 1]

    # Final answer: does entire s match entire p?
    return dp[0][0]
```

```python
def visualize_isMatch(s: str, p: str) -> bool:
    m, n = len(s), len(p)

    dp = [[False] * (n + 1) for _ in range(m + 1)]
    dp[m][n] = True

    print("Initial DP Table:")
    print_dp(dp, s, p)
    print("\n--------------------------------------\n")

    for i in range(m, -1, -1):
        for j in range(n - 1, -1, -1):

            match = (i < m) and (p[j] == s[i] or p[j] == ".")

            if j + 1 < n and p[j + 1] == "*":
                skip_star = dp[i][j + 2]
                use_star = match and dp[i + 1][j]
                dp[i][j] = skip_star or use_star
            else:
                dp[i][j] = match and dp[i + 1][j + 1]

            print(f"After computing dp[{i}][{j}] (s[{i}:] vs p[{j}:]):")
            print_dp(dp, s, p)
            print("--------------------------------------")

    print("\nFinal DP Table:")
    print_dp(dp, s, p)
    print("\nResult:", dp[0][0])
    return dp[0][0]


def print_dp(dp, s, p):
    # Column header
    print("        ", end="")
    for j in range(len(p)):
        print(f"{p[j]:>4}", end="")
    print(f"{'ε':>4}")

    # Rows
    for i, row in enumerate(dp):
        row_label = s[i] if i < len(s) else "ε"
        print(f"{row_label:>4}   ", end="")
        for val in row:
            cell = "T" if val else "F"
            print(f"{cell:>4}", end="")
        print()
```
