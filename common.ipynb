```python
from collections import deque, Counter, defaultdict
import heapq

# Sample data
students = [
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 78),
    ("David", 95),
    ("Eve", 88)
]

# 1. List operations
student_names = [name for name, _ in students]  # Extract names

# Append a new student
student_names.append("Frank")

# Insert at specific position
student_names.insert(2, "Grace")  # Insert Grace at index 2

# Remove an element by value
student_names.remove("Charlie")

# Pop last element
last_student = student_names.pop()

# Access by index
first_student = student_names[0]

print("List operations:")
print("Names after append, insert, remove, pop:", student_names)
print("Last student popped:", last_student)
print("First student:", first_student)
print()

# 2. Tuple operations
# Tuples are immutable, so common usage:
record = ("Heidi", 90)
# Access tuple items
name = record[0]
score = record[1]
print("Tuple operations:")
print(f"Name: {name}, Score: {score}")
print()

# 3. Dictionary operations
student_scores = {name: score for name, score in students}

# Add or update an entry
student_scores["Grace"] = 82

# Access value by key
grace_score = student_scores.get("Grace")  # safer than []
# Check if key exists
has_bob = "Bob" in student_scores

# Remove an entry
removed_score = student_scores.pop("Alice", None)  # returns None if key missing

print("Dictionary operations:")
print("Scores:", student_scores)
print("Grace's score:", grace_score)
print("Bob is present:", has_bob)
print("Removed Alice's score:", removed_score)
print()

# 4. Set operations
unique_students = set(student_scores.keys())

# Add a new item
unique_students.add("Ivy")

# Remove an item (raises error if missing)
unique_students.remove("Grace")

# Discard an item (no error if missing)
unique_students.discard("NotInSet")

# Membership test
has_david = "David" in unique_students

# Set operations
set_a = {"Alice", "Bob", "Charlie"}
set_b = {"Charlie", "David", "Eve"}

union_set = set_a.union(set_b)       # or set_a | set_b
intersection_set = set_a.intersection(set_b)  # or set_a & set_b
difference_set = set_a.difference(set_b)      # or set_a - set_b

print("Set operations:")
print("Unique students:", unique_students)
print("Has David:", has_david)
print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference:", difference_set)
print()

# 5. Deque operations
waiting_queue = deque(["Alice", "Bob", "Charlie"])

# Append to right (tail)
waiting_queue.append("David")

# Append to left (head)
waiting_queue.appendleft("Eve")

# Pop from right
last = waiting_queue.pop()

# Pop from left
first = waiting_queue.popleft()

print("Deque operations:")
print("Queue after appends:", waiting_queue)
print("Popped from right:", last)
print("Popped from left:", first)
print()

# 6. Heap operations
score_heap = []
# Push elements
heapq.heappush(score_heap, (85, "Alice"))
heapq.heappush(score_heap, (92, "Bob"))
heapq.heappush(score_heap, (78, "Charlie"))

# Pop smallest element
lowest_score = heapq.heappop(score_heap)

# Peek smallest element (without popping)
if score_heap:
    smallest = score_heap[0]
else:
    smallest = None

print("Heap operations:")
print("Lowest score popped:", lowest_score)
print("Smallest remaining element:", smallest)
print()

# 7. Counter operations
scores = ["A", "B", "A", "C", "B", "A"]
score_counter = Counter(scores)

# Count of 'A'
count_a = score_counter["A"]

# Most common elements
top_two = score_counter.most_common(2)

# Update counts
score_counter.update(["B", "C", "C"])

print("Counter operations:")
print("Counts:", score_counter)
print("Count of 'A':", count_a)
print("Top two:", top_two)
print()

# 8. DefaultDict operations
grouped = defaultdict(list)

# Add items to defaultdict
grouped["80s"].append("Alice")
grouped["90s"].append("Bob")
grouped["80s"].append("Eve")

# Access missing key returns default value (empty list)
print("DefaultDict operations:")
print("Grouped students:", dict(grouped))
print("Access missing key '70s':", grouped["70s"])  # returns []
print()
```
